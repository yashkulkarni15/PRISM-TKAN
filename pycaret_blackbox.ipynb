```python
from autogluon.tabular import TabularDataset, TabularPredictor

data_root = 'https://autogluon.s3.amazonaws.com/datasets/Inc/'
train_data = TabularDataset(data_root + 'train.csv')
test_data = TabularDataset(data_root + 'test.csv')

predictor = TabularPredictor(label='class').fit(train_data)
predictions = predictor.predict(test_data)
```

```python
predictor.leaderboard(test_data, silent=True)
```

```python
from pycaret.classification import *

# Setup
s = setup(df, target='diagnosis', ignore_features=['id'], session_id=123)

# Train model
black_box_model = create_model('dt')

# ===== SHAP (Global) =====
print("SHAP: Overall Feature Importance")
interpret_model(black_box_model)

# ===== LIME (Local) =====
print("LIME: Explaining Observation 0")
interpret_model(black_box_model, plot='reason', observation=0)

print("LIME: Explaining Observation 25")
interpret_model(black_box_model, plot='reason', observation=25)
```
